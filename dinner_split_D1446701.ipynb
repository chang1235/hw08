## 簡短題目說明
整理聚餐原始文字資料（去除頭尾空白、轉為小寫、移除 `NT$` 與逗號並轉換為數值型別），依規定**先以原始餐費計算 10% 服務費**，再扣除折價券得到應付總額，除以人數後計算每人分攤費用，最後判斷是否超過每人 `300` 元預算並輸出費用摘要。下方分別提供「測試一」「測試二」「測試三」三組測試，每次皆從原始資料出發只修改指定欄位並重新執行。

## 測試一（原始資料：4 人、折價券 100 元）

```python
# 測試一：原始輸入資料（4 人、折價券 100 元）
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"
service_rate = 0.10
discount = 100
budget_per_person = 300

# 1. 整理聚餐名稱（去頭尾空白並轉小寫）
clean_name = event_name.strip().lower()

# 2. 清理餐費文字與轉換人數型別（保留原始變數）
clean_meal_text = meal_amount.replace("NT$", "").replace(",", "").strip()
meal_price = float(clean_meal_text)
num_people = int(people)

# 3. 分步計算費用（先算服務費，再扣折價券，最後平分）
service_fee = meal_price * service_rate
total_amount = meal_price + service_fee - discount
per_person = total_amount / num_people

# 4. 判斷是否超過每人預算（布林值）
is_over_budget = per_person > budget_per_person

# 5. 輸出費用摘要
print("=== 測試一 ===")
print(f"聚餐名稱：{clean_name}")
print(f"原始餐費：{meal_price:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{num_people} 人")
print(f"每人費用：{per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
```
